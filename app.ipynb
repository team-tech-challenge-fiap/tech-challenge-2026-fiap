```python
import matplotlib.pyplot as plt

#colunas = []
    #col for col in df.columns
    #if col.startswith(("3_wpi_", "3_sss_"))
#]

colunas_wpi = [
"3_wpi_cei", "3_wpi_ced", "3_wpi_bsi", "3_wpi_bsd", "3_wpi_bii", "3_wpi_bid", "3_wpi_ni",  "3_wpi_nd",  "3_wpi_psi", "3_wpi_psd", "3_wpi_pii", "3_wpi_pid", "3_wpi_mi",  "3_wpi_md",  "3_wpi_tx",  "3_wpi_abd", "3_wpi_cue", "3_wpi_es",  "3_wpi_ei"
]

colunas_sss = [
"3_sss_f", "3_sss_s", "3_sss_tc", "3_sss_1", "3_sss_6", "3_sss_10", "3_sss_11", "3_sss_17", "3_sss_18", "3_sss_23", "3_sss_36"
]

cores = {
    0: "red",   # Fibromialgia
    1: "blue"   # Controle saudável
}

fig, ax = plt.subplots(figsize=(12, 6))

for grupo, nome in [(0, "Fibromialgia"), (1, "Controle")]:
    dados = df[df["gp"] == grupo

    for wpi in colunas_wpi:
        for sss in colunas_sss:

            ax.scatter(
                x=dados[wpi],
                y=dados[sss],
                color=cores[grupo],
                alpha=0.3,
                s=30
            )

ax.set_xlabel("WPI - Índice de Dor Generalizada")
ax.set_ylabel("SSS - Gravidade dos Sintomas")
ax.set_title("Distribuição de Fibromialgia - WPI x SSS")
ax.legend()

plt.tight_layout()
plt.show()
```
