```python
from pathlib import Path
import pandas as pd
import sys

project_root = Path.cwd()
if not (project_root / "src").exists():
    project_root = project_root.parent

sys.path.insert(0, str(project_root / "src"))

from cardio.features import add_features, get_feature, categorizar_pressao


```

```python
df = pd.read_csv("../data/raw/cardio_train.csv", sep=";")

df["age"] = (df["age"] / 365.25).astype(int)
df["imc"] = df["weight"] / (df["height"] / 100) ** 2


dados_invalidos = (
    (df["ap_hi"] <= 0) |
    (df["ap_lo"] <= 0) |
    (df["ap_lo"] >= df["ap_hi"]) |
    (df["ap_hi"] >= 300) |
    (df["ap_lo"] >= 200) |
    (df["ap_hi"] < 70) |
    (df["ap_lo"] < 40)
)

df_clean = df.loc[~dados_invalidos].copy()
```

```python

"""
Categoria de pressão arterial:

Classificação baseada na diretriz ACC/AHA de 2017.

• Normal: Sistólica < 120 e Diastólica < 80
• Elevada: Sistólica 120–129 e Diastólica < 80
• Hipertensão Estágio 1: Sistólica 130–139 ou Diastólica 80–89
• Hipertensão Estágio 2: Sistólica ≥ 140 ou Diastólica ≥ 90

Quando as pressões pertencem a categorias diferentes,
é considerada a categoria mais alta.

Esta variável é utilizada como atributo derivado
para Machine Learning, não como diagnóstico clínico.
"""

df_features = add_features(df_clean)

feat_A = df_features[get_feature("A")]
feat_B = df_features[get_feature("B")]

y=df_features["cardio"]


print("Conjunto A:", feat_A.shape)
print("Conjunto B:", feat_B.shape)

display(df_features.head())


display(
    df_features[
        ["imc", "pressao_pulso", "pam",
         "categoria_pressao", "faixa_imc"]
    ].head(10)
)
```
